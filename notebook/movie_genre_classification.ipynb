## 🔤 TF-IDF Feature Extraction

TF-IDF (Term Frequency-Inverse Document Frequency) is used to convert movie plot text into numerical features that can be processed by machine learning algorithms.

The vectorizer uses unigrams and bigrams while removing common English stop words.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

print("Models imported successfully!")
```

```python
train_path = "../data/train_data.txt"
test_path = "../data/test_data.txt"

train_df = pd.read_csv(
    train_path,
    sep=":::",
    engine="python",
    header=None,
    names=["id", "title", "genre", "plot"],
    on_bad_lines="skip"
)

test_df = pd.read_csv(
    test_path,
    sep=":::",
    engine="python",
    header=None,
    names=["id", "title", "genre", "plot"],
    on_bad_lines="skip"
)

# Convert text columns to strings
for df in [train_df, test_df]:
    df["title"] = df["title"].fillna("").astype(str).str.strip()
    df["genre"] = df["genre"].fillna("").astype(str).str.strip()
    df["plot"] = df["plot"].fillna("").astype(str).str.strip()

print("Training dataset shape:", train_df.shape)
print("Testing dataset shape:", test_df.shape)

print("\nFirst 5 rows:")
print(train_df.head())
```

```python
print("Columns:")
print(train_df.columns.tolist())

print("\nFirst row:")
print(train_df.iloc[0])

print("\nGenre examples:")
print(train_df["genre"].head(10).tolist())
```

```python
print("Missing values:")
print(train_df.isnull().sum())

print("\nTest dataset missing values:")
print(test_df.isnull().sum())
```

```python
# Remove missing values from important columns
train_df = train_df.dropna(subset=["plot", "genre"])
test_df = test_df.dropna(subset=["plot", "genre"])

# Convert text to lowercase
train_df["plot"] = train_df["plot"].str.lower()
test_df["plot"] = test_df["plot"].str.lower()

print("Training shape after cleaning:", train_df.shape)
print("Testing shape after cleaning:", test_df.shape)
```

```python
X_train = train_df["plot"]
y_train = train_df["genre"]

X_test = test_df["plot"]
y_test = test_df["genre"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

## 🔤 TF-IDF Feature Extraction

TF-IDF converts movie plot summaries into numerical features that machine learning models can understand.

We use unigrams and bigrams and remove common English stop words.

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

print("Models imported successfully!")
```

```python
naive_bayes_model = MultinomialNB()

naive_bayes_model.fit(X_train_tfidf, y_train)

print("Naive Bayes training completed!")
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_tfidf, y_train)

print("Logistic Regression training completed!")
```

```python
svm_model = LinearSVC(
    max_iter=2000,
    random_state=42
)

svm_model.fit(X_train_tfidf, y_train)

print("SVM training completed!")
```

```python
nb_pred = naive_bayes_model.predict(X_test_tfidf)
logistic_pred = logistic_model.predict(X_test_tfidf)
svm_pred = svm_model.predict(X_test_tfidf)

print("Predictions completed!")
```

```python
comparison_df = pd.DataFrame({
    "Actual Genre": y_test.values[:20],
    "Naive Bayes": nb_pred[:20],
    "Logistic Regression": logistic_pred[:20],
    "SVM": svm_pred[:20]
})

comparison_df
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate_model(name, y_true, y_pred):
    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(
            y_true, y_pred,
            average="weighted",
            zero_division=0
        ),
        "Recall": recall_score(
            y_true, y_pred,
            average="weighted",
            zero_division=0
        ),
        "F1-Score": f1_score(
            y_true, y_pred,
            average="weighted",
            zero_division=0
        )
    }

results = [
    evaluate_model("Naive Bayes", y_test, nb_pred),
    evaluate_model("Logistic Regression", y_test, logistic_pred),
    evaluate_model("SVM", y_test, svm_pred)
]

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "F1-Score",
    ascending=False
)

results_df
```

```python
best_model_name = results_df.iloc[0]["Model"]

print("Best Model:", best_model_name)
print("Best F1-Score:", results_df.iloc[0]["F1-Score"])
```
